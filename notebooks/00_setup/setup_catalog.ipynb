# Catalog and Schema Setup

This notebook initializes the Databricks catalog and schemas required for the FMCG Lakehouse project.  
It sets up a consistent namespace for Bronze, Silver, and Gold layers before any data ingestion or transformation occurs.

This step is a one-time setup and is a prerequisite for all downstream dimension and fact processing notebooks.

## Create Catalog

The catalog acts as the top-level container for all schemas and tables used in this project.


```python
%sql
CREATE CATALOG IF NOT EXISTS fmcg;
```

## Set Active Catalog

All schemas and tables created in this project will reside under the `fmcg` catalog.


```python
%sql
USE catalog fmcg;
```

## Create Gold Schema

The Gold schema stores analytics-ready, business-curated tables designed for reporting and BI consumption.


```python
%sql
CREATE SCHEMA IF NOT EXISTS fmcg.gold;
```

```python
%sql
SHOW DATABASES FROM fmcg;
```

## Create Bronze and Silver Schemas

- **Bronze**: Stores raw ingested data with minimal transformation  
- **Silver**: Stores cleaned and standardized data after applying business rules  

These schemas support the Medallion Architecture used throughout the pipeline.


```python
%sql
CREATE SCHEMA IF NOT EXISTS fmcg.bronze;
CREATE SCHEMA IF NOT EXISTS fmcg.silver;
```
